```python
import os

# Create fake mmcv package inside EMCAD folder
mmcv_base = "/kaggle/working/EMCAD/mmcv"
paths = [
    f"{mmcv_base}",
    f"{mmcv_base}/runner",
    f"{mmcv_base}/cnn",
    f"{mmcv_base}/cnn/utils",
    f"{mmcv_base}/cnn/utils/weight_init",
    f"{mmcv_base}/cnn/bricks",
    f"{mmcv_base}/utils",
    f"{mmcv_base}/ops",
]
for p in paths:
    os.makedirs(p, exist_ok=True)

# mmcv/__init__.py
with open(f"{mmcv_base}/__init__.py", "w") as f:
    f.write("")

# mmcv/runner/__init__.py — load_checkpoint
with open(f"{mmcv_base}/runner/__init__.py", "w") as f:
    f.write('''
import torch
import warnings

def load_checkpoint(model, filename, map_location="cpu", strict=False, logger=None):
    """Load checkpoint from file."""
    checkpoint = torch.load(filename, map_location=map_location, weights_only=False)
    
    if isinstance(checkpoint, dict):
        if "state_dict" in checkpoint:
            state_dict = checkpoint["state_dict"]
        elif "model" in checkpoint:
            state_dict = checkpoint["model"]
        else:
            state_dict = checkpoint
    else:
        state_dict = checkpoint
    
    # Remove 'module.' prefix
    new_state_dict = {}
    for k, v in state_dict.items():
        if k.startswith("module."):
            new_state_dict[k[7:]] = v
        else:
            new_state_dict[k] = v
    
    missing, unexpected = model.load_state_dict(new_state_dict, strict=strict)
    if missing and logger:
        logger.warning(f"Missing keys: {missing}")
    if unexpected and logger:
        logger.warning(f"Unexpected keys: {unexpected}")
    
    return checkpoint
''')

# mmcv/cnn/__init__.py — ConvModule, build_norm_layer
with open(f"{mmcv_base}/cnn/__init__.py", "w") as f:
    f.write('''
import torch.nn as nn

def build_norm_layer(cfg, num_features):
    """Build normalization layer."""
    if cfg is None:
        return nn.Identity(), None
    if isinstance(cfg, dict):
        cfg_type = cfg.get("type", "BN")
        if cfg_type in ("BN", "BN2d"):
            return nn.BatchNorm2d(num_features), cfg_type
        elif cfg_type == "SyncBN":
            return nn.BatchNorm2d(num_features), cfg_type
        elif cfg_type == "LN":
            return nn.LayerNorm(num_features), cfg_type
        elif cfg_type == "GN":
            num_groups = cfg.get("num_groups", 32)
            return nn.GroupNorm(num_groups, num_features), cfg_type
    return nn.BatchNorm2d(num_features), "BN"

class ConvModule(nn.Module):
    """Simplified ConvModule compatible with PyTorch 2.x"""
    def __init__(self, in_channels, out_channels, kernel_size, stride=1, padding=0,
                 dilation=1, groups=1, bias=False, norm_cfg=None, act_cfg=None,
                 inplace=True, order=("conv", "norm", "act"), **kwargs):
        super().__init__()
        self.order = order
        self.with_norm = norm_cfg is not None
        self.with_activation = act_cfg is not None
        
        self.conv = nn.Conv2d(
            in_channels, out_channels, kernel_size,
            stride=stride, padding=padding, dilation=dilation,
            groups=groups, bias=bias
        )
        
        if self.with_norm:
            self.norm_name, norm = build_norm_layer(norm_cfg, out_channels)
            self.add_module(self.norm_name, norm)
        
        if self.with_activation:
            act_type = act_cfg.get("type", "ReLU") if isinstance(act_cfg, dict) else "ReLU"
            if act_type == "ReLU":
                self.activate = nn.ReLU(inplace=inplace)
            elif act_type == "GELU":
                self.activate = nn.GELU()
            elif act_type == "SiLU" or act_type == "Swish":
                self.activate = nn.SiLU()
            else:
                self.activate = nn.ReLU(inplace=inplace)
    
    @property
    def norm(self):
        if self.with_norm:
            return getattr(self, self.norm_name)
        return None
    
    def forward(self, x):
        for layer in self.order:
            if layer == "conv":
                x = self.conv(x)
            elif layer == "norm" and self.with_norm:
                x = self.norm(x)
            elif layer == "act" and self.with_activation:
                x = self.activate(x)
        return x

def constant_init(module, val, bias=0):
    if hasattr(module, "weight") and module.weight is not None:
        nn.init.constant_(module.weight, val)
    if hasattr(module, "bias") and module.bias is not None:
        nn.init.constant_(module.bias, bias)

def kaiming_init(module, a=0, mode="fan_out", nonlinearity="relu", bias=0, distribution="normal"):
    if hasattr(module, "weight") and module.weight is not None:
        nn.init.kaiming_normal_(module.weight, a=a, mode=mode, nonlinearity=nonlinearity)
    if hasattr(module, "bias") and module.bias is not None:
        nn.init.constant_(module.bias, bias)
''')

# mmcv/cnn/utils/__init__.py
with open(f"{mmcv_base}/cnn/utils/__init__.py", "w") as f:
    f.write("from .weight_init import trunc_normal_, constant_init, kaiming_init\n")

# mmcv/cnn/utils/weight_init/__init__.py
with open(f"{mmcv_base}/cnn/utils/weight_init/__init__.py", "w") as f:
    f.write('''
import torch.nn as nn

def trunc_normal_(tensor, mean=0., std=1., a=-2., b=2.):
    return nn.init.trunc_normal_(tensor, mean=mean, std=std, a=a, b=b)

def constant_init(module, val, bias=0):
    if hasattr(module, "weight") and module.weight is not None:
        nn.init.constant_(module.weight, val)
    if hasattr(module, "bias") and module.bias is not None:
        nn.init.constant_(module.bias, bias)

def kaiming_init(module, a=0, mode="fan_out", nonlinearity="relu", bias=0, distribution="normal"):
    if hasattr(module, "weight") and module.weight is not None:
        nn.init.kaiming_normal_(module.weight, a=a, mode=mode, nonlinearity=nonlinearity)
    if hasattr(module, "bias") and module.bias is not None:
        nn.init.constant_(module.bias, bias)
''')

# mmcv/cnn/bricks/__init__.py
with open(f"{mmcv_base}/cnn/bricks/__init__.py", "w") as f:
    f.write('''
import torch.nn as nn

class DropPath(nn.Module):
    """Drop paths (Stochastic Depth) per sample."""
    def __init__(self, drop_prob=0.):
        super().__init__()
        self.drop_prob = drop_prob
    
    def forward(self, x):
        if self.drop_prob == 0. or not self.training:
            return x
        keep_prob = 1 - self.drop_prob
        shape = (x.shape[0],) + (1,) * (x.ndim - 1)
        random_tensor = keep_prob + torch.rand(shape, dtype=x.dtype, device=x.device)
        random_tensor.floor_()
        output = x.div(keep_prob) * random_tensor
        return output
''')

# mmcv/utils/__init__.py — Registry
with open(f"{mmcv_base}/utils/__init__.py", "w") as f:
    f.write('''
class Registry:
    """Simplified registry."""
    def __init__(self, name):
        self.name = name
        self._module_dict = {}
    
    def register_module(self, name=None, module=None, force=False):
        def _register(cls):
            module_name = name if name is not None else cls.__name__
            if module_name in self._module_dict and not force:
                raise KeyError(f"{module_name} already registered")
            self._module_dict[module_name] = cls
            return cls
        if module is not None:
            return _register(module)
        return _register
    
    def get(self, key):
        return self._module_dict.get(key)
    
    def __contains__(self, key):
        return key in self._module_dict
''')

# mmcv/ops/__init__.py
with open(f"{mmcv_base}/ops/__init__.py", "w") as f:
    f.write("")

print("✅ mmcv shim created successfully!")
print("Files created:")
for root, dirs, files in os.walk(mmcv_base):
    for f in files:
        print(f"  {os.path.join(root, f)}")
```

```python
import os
import shutil
from sklearn.model_selection import train_test_split


src="/kaggle/working/EMCAD/data/polyp/Kvasir-SEG"


images=sorted(os.listdir(src+"/images"))


train_imgs, temp_imgs = train_test_split(
    images,
    test_size=0.3,
    random_state=42
)


val_imgs, test_imgs = train_test_split(
    temp_imgs,
    test_size=0.5,
    random_state=42
)


splits={
    "train":train_imgs,
    "val":val_imgs,
    "test":test_imgs
}


for split, files in splits.items():

    img_dir=f"{src}/{split}/images"
    mask_dir=f"{src}/{split}/masks"

    os.makedirs(img_dir, exist_ok=True)
    os.makedirs(mask_dir, exist_ok=True)


    count=0

    for img in files:

        img_path=f"{src}/images/{img}"

        # image copy
        shutil.copy(
            img_path,
            f"{img_dir}/{img}"
        )


        # get mask name automatically
        name=os.path.splitext(img)[0]

        mask_candidates=[
            name+".png",
            name+".jpg",
            name+".jpeg"
        ]


        mask_found=False

        for m in mask_candidates:
            if os.path.exists(f"{src}/masks/{m}"):

                shutil.copy(
                    f"{src}/masks/{m}",
                    f"{mask_dir}/{m}"
                )

                mask_found=True
                break


        if mask_found:
            count+=1
        else:
            print("Mask missing:", img)


    print(split, "images:", len(files), "masks:", count)


print("DONE")
```

```python
import os
import cv2
import matplotlib.pyplot as plt
from glob import glob


image_dir = "/kaggle/working/EMCAD/data/polyp/Kvasir-SEG/test/images"
mask_dir = "/kaggle/working/EMCAD/data/polyp/Kvasir-SEG/test/masks"

pred_dir = "/kaggle/working/EMCAD/predictions_polyp/Kvasir-SEG_pvt_v2_b2_EMCAD_kernel_sizes_1_3_5_dw_parallel_add_lgag_ks_3_ef2_act_mscb_relu6_bs8_cas_lr0.0005_e5_augTrue_run1_t213833/Kvasir-SEG/test"


# sorted files
images = sorted(glob(image_dir+"/*"))
masks = sorted(glob(mask_dir+"/*"))
preds = sorted(glob(pred_dir+"/*"))


print("Images:", len(images))
print("Masks:", len(masks))
print("Predictions:", len(preds))


plt.figure(figsize=(15,15))


for i in range(5):

    img = cv2.imread(images[i])
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    mask = cv2.imread(masks[i],0)

    pred = cv2.imread(preds[i],0)


    plt.subplot(5,3,i*3+1)
    plt.imshow(img)
    plt.title("Original Image")
    plt.axis("off")


    plt.subplot(5,3,i*3+2)
    plt.imshow(mask,cmap="gray")
    plt.title("Ground Truth")
    plt.axis("off")


    plt.subplot(5,3,i*3+3)
    plt.imshow(pred,cmap="gray")
    plt.title("EMCAD Prediction")
    plt.axis("off")


plt.tight_layout()
plt.show()
```
